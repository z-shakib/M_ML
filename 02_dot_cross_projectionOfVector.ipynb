```python
import numpy as np 
```

```python
# dot product of two vector 

vector_a = np.array([2,3])
vector_b = np.array([4,4]) 

a_dot_b = np.dot(vector_a, vector_b) 
a_dot_b 
```

```python

vector_c = np.array([40, 20, 35])
vector_d = np.array([53, 24, 68])

c_dot_d = np.dot(vector_c, vector_d)
c_dot_d 
```

```python
# cross product of two vector: 

a_cross_b = np.cross(vector_a, vector_b) 
a_cross_b 
```

```python
c_cross_d = np.cross(vector_c, vector_d) 
c_cross_d 
```

```python
# projection of vector c on vector d 
```
