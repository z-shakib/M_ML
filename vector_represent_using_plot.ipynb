```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
sns.set() 
```

```python
# represent a vector

plt.quiver(0,0,4,5)
plt.show() 
```

```python
plt.quiver(0,0,4,5, scale_units='xy', angles='xy', scale=1, color='b')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show() 
```

```python
# show 2 vector

plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,3,-2, scale_units='xy', angles='xy', scale=1, color='y')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show() 
```

```python
# Vector Addition 

vector_a = np.asarray([0,0,2,3])
vector_b = np.asarray([0,0,3,-2]) 

sumOfVector = vector_a + vector_b 
sumOfVector
```

```python
# plot sum of two vector 

plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,3,-2, scale_units='xy', angles='xy', scale=1, color='y')
plt.quiver(0, 0, 5, 1, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show() 
```

```python
# Vector Subtraction 

diffrence = vector_a - vector_b 
diffrence 
```

```python
# represent vector subtraction 

plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0,0,3,-2, scale_units='xy', angles='xy', scale=1, color='y')
plt.quiver(0,  0, -1,  5, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show() 
```

```python
# multiplying a vector by scaler: 

vector_c = vector_a*2 

vector_c 
```

```python
plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0, 0, 4, 6, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show() 
```

```python
# multiply using negetive number 

plt.quiver(0,0,2,3, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(0, 0, -4, -6, scale_units='xy', angles='xy', scale=1, color='r')

plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show() 
```
